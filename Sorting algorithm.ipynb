```python
def my_sort(numbers):
    """Sort a list of ints by selection sort. Returns (sorted_list, comparisons)."""
    arr = numbers[:]
    n = len(arr)
    comparisons = 0

    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_index]:
                min_index = j
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr, comparisons


# --- two short tests ---
print(my_sort([5, 3, 8, 1, 9]))
print(my_sort([9, 8, 7, 6, 5]))


# --- one long test: n = 10 ---
data = [7, 2, 9, 4, 1, 8, 3, 6, 5, 0]
result, comparisons = my_sort(data)
print(result)             
print(comparisons)       
```
