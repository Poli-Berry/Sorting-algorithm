```python
import numpy as np

# 3 departments (rows) x 10 months Jan..Oct (columns)
income = np.random.randint(100_000, 500_000, size=(3, 10))
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct"]

print("Gross income (departments x months):")
print(income)
print("Columns:", months)


# Net income with 30% tax everywhere
tax_30 = 0.30
net_income = income * (1 - tax_30)

print("\nNet income after flat 30% tax:")
print(net_income)


# Tax changed: May -> 40%, August -> back to 30%
# So: May and June are taxed at 40% (May is when it started,
# and it "changed back" in August, meaning May, Jun, Jul are at 40%).
# Months at 40%: May (index 4), Jun (5), Jul (6)

tax_rate = np.full(10, tax_30)      # 30% everywhere
tax_rate[4:7] = 0.40                # May, Jun, Jul -> 40%

print("\nTax rate per month:", tax_rate)

# Recalculation: (3,10) * (10,) multiplies each column by its rate
net_income_recalc = income * (1 - tax_rate)

print("\nRecalculated net income:")
print(net_income_recalc)


# 2nd quarter = Apr, May, Jun -> indices 3, 4, 5
q2 = net_income_recalc[:, 3:6]

print("\n2nd quarter net income (Apr, May, Jun):")
print(q2)
```
